### Boolean masking and fancy indexing

```python
sales = np.array([45, 90, 30, 75, 20, 60, 85, 50])

mask = sales > 50
print(mask)                                 
print(sales[mask])                           
print(sales[(sales > 30) & (sales < 80)])    
```

```python
print(sales[[0, 3, 5]])   
print(b[b > 2])           
print(b[[1, 0]])            
print(b[[0, 1], [2, 0]])     
```

```python
sales[sales < 30] = 0
print(sales)               
```

### Statistical measures

```python
units = np.array([2, 5, 3, 6, 1, 4, 7, 3])
price = np.array([45, 90, 30, 75, 20, 60, 85, 50])

print(np.mean(price))                      
print(np.median(price))                    
print(np.std(price))                       
print(np.percentile(price, [25, 50, 75]))  
print(np.corrcoef(units, price))
print(np.corrcoef(units, price)[0, 1])     
```

### NumPy vs Python loop on 1 million rows

```python
import time

np.random.seed(0)
n = 1_000_000
qty = np.random.randint(1, 10, n)
unit_price = np.random.rand(n) * 100

qty_list, price_list = qty.tolist(), unit_price.tolist()    
```

##### Python loop

```python
t = time.perf_counter()
total_loop = 0
for i in range(n):
    total_loop += qty_list[i] * price_list[i]
loop_time = time.perf_counter() - t
```

##### NumPy

```python
t = time.perf_counter()
total_np = (qty * unit_price).sum()
np_time = time.perf_counter() - t
```

### Compare

```python
print(total_loop, total_np, np.isclose(total_loop, total_np))
print(f'Loop: {loop_time:.4f}s | NumPy: {np_time:.4f}s | NumPy is {loop_time/np_time:.0f}x faster')
```

# In this test, NumPy took 0.0066 seconds and the Python loop took 0.3268 seconds to calculate the
same total over 1,000,000 rows, so NumPy was about 50 times faster. Both methods gave the same
result. NumPy is faster because it applies the operation to the whole array at once using
vectorised code, while a loop handles one value at a time. This makes NumPy the better choice for
large datasets and numerical calculations.
