```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Data = [['Jan','North',20,4200,600], ['Feb','South',25,5100,750],
        ['Mar','East',22,4600,500], ['Apr','West',30,6200,980],
        ['May','North',28,5800,870], ['Jun','South',35,7100,1200],
        ['Jul','East',31,6400,900], ['Aug','West',40,8300,1500],
        ['Sep','North',38,7900,1300], ['Oct','South',45,9200,1700]]

df = pd.DataFrame(Data, columns=['Month','Region','Units','Sales','Profit'])
df.head()
```

### Line plot, with a callout


```python
plt.plot(df['Month'], df['Sales'], marker='o')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Month vs Sales')
plt.grid(True)

plt.annotate('Highest: 9200', xy=('Oct', 9200), xytext=('Jun', 9000),
             arrowprops=dict(arrowstyle='->'))

plt.savefig('line.png')
plt.show()
```

### Bar plot

```python
region_sales = df.groupby('Region')['Sales'].sum()

bars = plt.bar(region_sales.index, region_sales.values)
plt.bar_label(bars)
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.title('Sales by Region')

plt.savefig('bar.png')
plt.show()
```

### Scatter plot

```python
plt.scatter(df['Units'], df['Sales'])
plt.xlabel('Units')
plt.ylabel('Sales')
plt.title('Units vs Sales')

plt.savefig('scatter.png')
plt.show()
```

### Histogram

```python
plt.hist(df['Sales'], bins=5, edgecolor='white')
plt.xlabel('Sales')
plt.ylabel('Count')
plt.title('Sales Distribution')

plt.savefig('histogram.png')
plt.show()
```

### Box plot

```python
plt.boxplot(df['Sales'])
plt.ylabel('Sales')
plt.title('Sales Box Plot')

plt.savefig('box.png')
plt.show()
```

### Seaborn theme

```python
sns.set_theme(style='darkgrid', palette='muted', context='talk')

plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Month', y='Sales', marker='o')
plt.title('Month vs Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.savefig('seaborn_line.png')
plt.show()
```
