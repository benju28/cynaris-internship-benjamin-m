```python
import pandas as pd

sales = pd.DataFrame([[1,101,'North','Laptop',50000], [2,102,'South','Phone',20000],
                      [3,103,'North','Phone',18000],  [4,101,'East','Laptop',52000],
                      [5,104,'West','Tablet',15000],  [6,105,'South','Laptop',48000],
                      [7,102,'East','Tablet',14000],  [8,106,'West','Phone',22000]],
                     columns=['order_id','cust_id','region','product','amount'])

customers = pd.DataFrame([[101,'Arun'], [102,'Bala'], [103,'Charan'], [104,'Divya'], [107,'Esha']],
                         columns=['cust_id','name'])
```

### Merge with all join types

```python
inner = pd.merge(sales, customers, on='cust_id', how='inner')
inner
```

```python
left = pd.merge(sales, customers, on='cust_id', how='left')
left
```

```python
right = pd.merge(sales, customers, on='cust_id', how='right')
right
```

```python
outer = pd.merge(sales, customers, on='cust_id', how='outer')
outer
```

### Pivot table 

```python
pivot = pd.pivot_table(sales, index='region', columns='product',
                       values='amount', aggfunc='sum', fill_value=0)
pivot
```

### Melt, wide to long

```python
wide = pd.DataFrame([['North',100,120,130], ['South',90,95,110],
                     ['East',80,85,70],     ['West',60,75,90]],
                    columns=['region','Jan','Feb','Mar'])

long = pd.melt(wide, id_vars='region', var_name='month', value_name='sales')
long.head()
```

### Method chaining

```python
result = (
    sales
    .query('amount > 15000')
    .assign(amount_k=lambda d: d['amount'] / 1000)
    .groupby('region')['amount_k'].sum()
    .sort_values(ascending=False)
    .reset_index()
)
result
```
